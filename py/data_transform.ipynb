```python
import urllib.request

def read_tsv(link):

    headers = {
            "User-Agent": (
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                "(KHTML, like Gecko) Chrome/126.0 Safari/537.36"
            )
        }
    try:
        req = urllib.request.Request(link, headers=headers)
        with urllib.request.urlopen(req, timeout=30) as resp:
            text = resp.read()
            return text.decode("utf-8")
    except:
        print( "Not good" )

def format_data(data):
    # data = data.decode("utf-8")
    data = data.split("\n")
    data = [[n.replace("\r", "") for n in i.split("\t")] for i in data]
    return data
```

```python
PUBLISH_LINK = "https://docs.google.com/spreadsheets/d/e/2PACX-1vTximK54RzevgtA54SwKnbA7P8o_d_nPcaqr0VG1jh7q3bbkOQSgesrUoeALTWujpycf-AtXO-eXFxU/pub?output=tsv"

data = read_tsv( PUBLISH_LINK )
data
```

```python
import pandas as pd
def make_df(data_list):
    formatted_data = format_data(data_list)
    formatted_data = formatted_data[1:] #skip the first line because of the double headers
    columns = formatted_data[0]
    print( columns )
    df = pd.DataFrame( data = formatted_data[1:], columns = columns )
    df = df.replace("", None )
    # Filter to only necessary columns
    df = df[['Goal','Objective Code','Objective Desc','Subtitle','Action Code','Action Desc','Activities-Deliverables','Status','Lead Implementers']]

    return df

df = make_df( data )
df.head()
```

```python
cols_tofill = ['Goal','Objective Code','Objective Desc','Subtitle','Action Code','Action Desc']

for c in cols_tofill:
    df[c] = df[c].ffill()
    df[c] = df[c].str.strip()

df.head()
```
